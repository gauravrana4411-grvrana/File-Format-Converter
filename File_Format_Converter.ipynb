Importing Required Modules

```python
import os
import glob
import re 
import json
import pandas as pd
```

Extracting Column Names from Schema

```python
def get_column_names(schemas, ds_name, sorting_key = 'column_position'):
    columns_details = schemas[ds_name]
    sorted_columns = sorted(columns_details, key = lambda x: x[sorting_key])
    return [col['column_name'] for col in sorted_columns]
```

Reading CSV/Text Files

```python
def read_csv(file,schemas):
    file_path_list = re.split(r'[\\/]', file)
    ds_name = file_path_list[-2]
    columns = get_column_names(schemas,ds_name)
    df = pd.read_csv(file, names=columns)
    return df
```

Exporting Data into JSON Format

```python
def to_json(df, base_dir, ds_name, file_name):
    dir_file_path = f"{base_dir}/{ds_name}"
    os.makedirs(dir_file_path, exist_ok = True)
    json_file_path = f"{dir_file_path}/{file_name}"
    #print(json_file_path)
    df.to_json(json_file_path, orient = 'records', lines = True)
```

Dataset File Conversion Logic

```python
def file_converter(src_dir, base_dir, schemas, ds_name):
    
    files = glob.glob(f'{src_dir}/{ds_name}/*')

    for file in files:
        print(f"Processing {file}")
        df = read_csv(file, schemas)
        file_name = re.split(r'[\\/]', file)[-1]
        #print(file_name)
        to_json(df, base_dir, ds_name, file_name) 
```

Main Execution Logic

```python
def process_files(ds_name = None):
    src_dir = 'retail_db'
    base_dir = 'retail_db_json'

    schemas = json.load(open('schemas.json'))
    src_file_names=glob.glob('retail_db/*/*')
    
    if not ds_name:
        ds_name = schemas.keys()
    for ds_name in ds_name:
        file_converter(src_dir, base_dir, schemas, ds_name) 
        print("File Successfully created.")   

    
```
